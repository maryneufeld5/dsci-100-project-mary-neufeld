```r
#set appropriate graph size 
options(repr.plot.width = 9, repr.plot.height = 9)

#create proportions column for tourney finishes
basketball_proportion<-tidy_data|>
group_by(tourney_finish)|>
summarize(n=n())|>
mutate(proportion=n/sum(n))

# plot tourney finish proportions
basketball_bar_plot<-basketball_proportion|>
ggplot(aes(y=fct_reorder(tourney_finish, proportion, .desc=TRUE), x=proportion))+
geom_bar(stat = "identity", fill="steelblue")+
# label the axis and give title
labs(x="Proportion", y="Tournament Finish", title="Tourney Finish Proportions for NCAA Division I Women’s Basketball" )+
# adjust text size to make it more readable
theme(text=element_text(size=13))

basketball_bar_plot

```

Visualization 1: My first scatterplot helps visualize the relationship between conference win percentages and regional win percentages for teams in the NCAA's Div I Women's Basketball. The data points are also coloured by tournament finish. The plot shows a positive relationship between conference and regional win percentages and also highlights that teams that progress further in the tournament often have higher conference and regional win percentages as well. This is relevant to the analysis as it helps us identify that tournament finish (categorical variable) should be able to be predicted by both regional and conference win percentages when using KNN classification later on. 

```r
#set appropriate graph size 
options(repr.plot.width = 10, repr.plot.height = 9)

#create scatterplot to compare conference win percentages and regional win percentages for different tournament finishes
basketball_scatterplot<-tidy_data|>
ggplot(aes(x=reg_wins_pct, y=conf_wins_pct, colour=tourney_finish))+
       geom_point()+
       labs(x="Regional Wins \n (%)", y="Conference Wins \n (%)", colour="Tournament Finish", title="Conference Win Percentages vs. Regional Win Percentages")

       basketball_scatterplot
```

Visualization 2: My bar plot helps to visualize how average regional win percentages differ for the varying tournament finishing spots in NCAA's Div I Women's Basketball. The plot shows that, in most cases, higher placing in the tournament is related to a higher average win percentage in the regular season. This is relevant to the analysis as it helps us to see that regional win percentage should be a good classifier for tournament finish in our KNN classification later on. 

```r
#set appropriate graph size 
options(repr.plot.width = 10, repr.plot.height = 9)

# create column of mean regional win percentages for diff tourney finishes
bar_tidy_data<-tidy_data|>
group_by(tourney_finish)|>
summarize(mean_reg_wins_pct=mean(reg_wins_pct))

#create bar plot to compare mean regional win percentages for tourney finishes
basketball_barplot<-bar_tidy_data|>
ggplot(aes(y=fct_reorder(tourney_finish, mean_reg_wins_pct, .desc=FALSE), x=mean_reg_wins_pct))+
geom_bar(stat="identity", fill="steelblue")+
labs(x="Mean Regional Wins \n (%)", y="Tournament Finish", title="Average Regional Win Percentage by Tournament Finish")+
# adjust text size to make it more readable
theme(text=element_text(size=13))

basketball_barplot
```
